## Module map

| Module | Covers |
|---|---|
| `varistar.timeseries` | `TimeSeries` — I/O, cleaning, stats |
| `varistar.lightcurve` | `LightCurve` — period search, folding, fitting |
| `varistar.groups` | `TestGroup` — batch workflows |
